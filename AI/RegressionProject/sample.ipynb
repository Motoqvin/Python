## Task 1

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('rost_ves.csv')

height = data['Рост (см)'].tolist()
weight = data['Вес (кг)'].tolist()
```

```python
plt.scatter(height, weight, color='green')

plt.xlabel('Рост (см)')
plt.ylabel('Вес (кг)')
plt.title('Зависимость веса от роста')

plt.show()
```

```python
hours = [0, 4, 8, 12, 16, 20, 24]
temp = [15, 14, 18, 22, 25, 20, 17]

plt.plot(hours, temp, color='red')

plt.xlabel('Часы')
plt.ylabel('Температура')
plt.title('Температура в течение суток')

plt.grid(True)

plt.show()
```
