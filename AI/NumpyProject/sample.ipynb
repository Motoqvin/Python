## Task 1

```python
import numpy as np
```

```python
arr = np.random.randint(1, 51, size=(4, 4))
```

```python
for i in arr:
    for j in i:
        print(j, end=" | ")
    print(end="\n")
    print("-----------------")
```

```python
col_sums = arr.sum(axis=0)
row_sums = arr.sum(axis=1)
```

```python
mean_val = arr.mean()
```

```python
print("Row sums:", row_sums)
print("Col sums:", col_sums)
print("General mean value:", mean_val)
```

### Отчет по продажам

В этом разделе мы считаем итоги по простым данным

```python
sales = np.array([120, 135, 100, 150, 90, 200, 170])
```

```python
sales_sum = sales.sum()
average_sales = sales.mean()
max_val = sales.max()
min_val = sales.min()

print(sales_sum)
print(average_sales)
print(max_val)
print(min_val)
```
