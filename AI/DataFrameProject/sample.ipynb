## Task 1

```python
import pandas as pd
```

```python
books = pd.DataFrame({
    'Название': ['Python 101', 'Data Science', 'AI Revolution', 'Deep Learning', 'Flask Web', 'Clean Code'],
    'Автор': ['John Doe', 'Anna Smith', 'Elon Data', 'Mike Lee', 'Sara Dev', 'Robert Martin'],
    'Год издания': [2010, 2018, 2020, 2015, 2021, 2008],
    'Цена': [25.5, 30.0, 45.0, 35.0, 40.0, 28.0]
})

print(books)
```

```python
avg_price = books['Цена'].mean()
print(avg_price)
```

```python
books_after_date = books[books['Год издания'] > 2015]
print(books_after_date)
```

```python
sorted_books = books.sort_values(by='Цена')
print(sorted_books)
```

## Task 3

```python
food = pd.DataFrame({
    'Продукт': ['Яблоко', 'Бургер', 'Салат', 'Пицца', 'Курица', 'Шоколад', 'Овсянка', 'Йогурт', 'Орехи', 'Сок'],
    'Категория': ['Фрукты', 'Фастфуд', 'Овощи', 'Фастфуд', 'Мясо', 'Сладости', 'Зерновые', 'Молочные', 'Орехи', 'Напитки'],
    'Калории': [52, 540, 120, 800, 250, 600, 150, 180, 650, 90],
    'Белки': [0.3, 25, 2, 30, 27, 5, 6, 10, 20, 1]
})

print(food)
```

```python
high_calory_food = food[food['Калории'] > 300]
print(high_calory_food)
```

```python
avg_protein = food.groupby('Категория')['Белки'].mean()
print(avg_protein)
```

```python
sorted_food = food.sort_values('Калории', ascending=False)
print(sorted_food)
```

## Task 4

```python
employees = pd.DataFrame({
    'Имя': ['Анна', 'Иван', 'Олег', 'Анна', 'Света', 'Иван', 'Марк', 'Олег'],
    'Проект': ['A', 'A', 'B', 'C', 'C', 'B', 'A', 'C'],
    'Часы': [10, 15, 20, 12, 18, 25, 8, 22]
})

print(employees)
```

```python
hours_employees = employees.groupby('Имя')['Часы'].max()
print(hours_employees)
```

```python
hours_projects = employees.groupby('Проект')['Часы'].max()
print(hours_projects)
```

```python
print(hours_employees.idxmax())
```

## Task 5

```python
tickets = pd.DataFrame({
    'Фильм': ['Avatar', 'Titanic', 'Inception', 'Matrix', 'Avatar', 'Titanic', 'Inception', 'Matrix', 'Avatar', 'Matrix', 'Titanic', 'Inception'],
    'Город': ['Баку', 'Баку', 'Москва', 'Москва', 'Киев', 'Киев', 'Баку', 'Минск', 'Минск', 'Москва', 'Киев', 'Баку'],
    'Проданные билеты': [500, 300, 250, 400, 450, 200, 220, 310, 390, 270, 260, 330]
})

print(tickets)
```

```python
tickets_film = tickets.groupby('Фильм')['Проданные билеты'].sum()
print(tickets_film)
```

```python
tickets_city = tickets.groupby('Город')['Проданные билеты'].sum()
print(tickets_city)
```

```python
top_film = tickets_film.idxmax()
print(top_film)
```
